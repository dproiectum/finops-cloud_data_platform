# 03 — Backfill des billings mensuels
Exécute la même clôture mensuelle sur une plage inclusive de mois.

```python
ENVIRONMENT = "dev"
START_MONTH = ""
END_MONTH = ""
ARCHIVE = "true"
try:
    dbutils.widgets.text("environment", ENVIRONMENT)
    dbutils.widgets.text("start_month", START_MONTH)
    dbutils.widgets.text("end_month", END_MONTH)
    dbutils.widgets.dropdown("archive", ARCHIVE, ["true", "false"])
    ENVIRONMENT = dbutils.widgets.get("environment")
    START_MONTH = dbutils.widgets.get("start_month")
    END_MONTH = dbutils.widgets.get("end_month")
    ARCHIVE = dbutils.widgets.get("archive")
except NameError:
    pass
if not START_MONTH or not END_MONTH:
    raise ValueError("Renseignez start_month et end_month au format YYYY-MM.")
```

```python
from finops_cloud.jobs.billing_backfill import run

results = run(
    ENVIRONMENT,
    START_MONTH,
    END_MONTH,
    archive=ARCHIVE.lower() == "true",
)
display(results)
```
